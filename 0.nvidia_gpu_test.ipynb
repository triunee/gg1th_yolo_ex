# GPU테스트

- CUDA 환경셋팅

- 간단 테스트

```python
import torch

print("CUDA 가능 여부:", torch.cuda.is_available())
print("사용 GPU:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "GPU 없음")
print("PyTorch CUDA 버전:", torch.version.cuda)
```

- 연산 테스트

```python
import torch

# 1️⃣ CUDA 사용 가능 여부 확인
print("CUDA available:", torch.cuda.is_available())

# 2️⃣ 사용 가능한 GPU 개수
print("GPU count:", torch.cuda.device_count())

# 3️⃣ 현재 사용 중인 GPU 이름
if torch.cuda.is_available():
    print("GPU name:", torch.cuda.get_device_name(0))
else:
    print("GPU is not available.")

# 4️⃣ 간단한 연산 테스트
if torch.cuda.is_available():
    # 랜덤 텐서를 GPU로 보내 연산 수행
    x = torch.rand(3, 3).to("cuda")
    y = torch.rand(3, 3).to("cuda")
    z = x + y
    print("GPU computation success! Result:\n", z)
else:
    print("Using CPU for computation.")
```
