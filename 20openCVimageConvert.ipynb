```python
import cv2
import numpy as np
```

```python
print("OpenCV version:")
print(cv2.__version__)
```

```python
img = cv2.imread("./images_opencv/ieu-01.jpg")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
# 원본 이미지
cv2.imshow("Original", img)

# 회색조 이미지
cv2.imshow("Gray", gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
