# 카메라 실시간테스트

## 카메라 사용 여부 체크

```python
import cv2

# cap = cv2.VideoCapture(0, cv2.CAP_V4L2)  # linux 용
cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)   # windows 용

print("카메라 열림:", cap.isOpened())

if cap.isOpened():
    ret, frame = cap.read()
    print("프레임 읽기:", ret)

cap.release()
```

## 카메라 켜기

```python
import cv2

# cap = cv2.VideoCapture(0, cv2.CAP_V4L2)  # linux 용
cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)   # windows 용

cap.set(
    cv2.CAP_PROP_FOURCC,
    cv2.VideoWriter_fourcc(*'YUYV')
)

cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)
cap.set(cv2.CAP_PROP_FPS, 30)

print("카메라 열림:", cap.isOpened())

print("실제 WIDTH :", cap.get(cv2.CAP_PROP_FRAME_WIDTH))
print("실제 HEIGHT:", cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
print("실제 FPS   :", cap.get(cv2.CAP_PROP_FPS))

ret, frame = cap.read()

print("프레임 읽기:", ret)

if ret:
    print("frame shape:", frame.shape)

cap.release()
```

## 실시간으로 카메라 켜기

```python
import cv2

cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)
cap.set(cv2.CAP_PROP_FOURCC, cv2.VideoWriter_fourcc(*'YUYV'))
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 640)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 480)
cap.set(cv2.CAP_PROP_FPS, 30)

print("카메라 열림:", cap.isOpened())
print("실제 WIDTH :", cap.get(cv2.CAP_PROP_FRAME_WIDTH))
print("실제 HEIGHT:", cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
print("실제 FPS   :", cap.get(cv2.CAP_PROP_FPS))

if not cap.isOpened():
    raise RuntimeError("카메라를 열 수 없습니다.")

try:
    while True:
        ret, frame = cap.read()
        if not ret:
            print("프레임을 읽을 수 없습니다.")
            break
        cv2.imshow("Webcam", frame)
        key = cv2.waitKey(1) & 0xFF
        if key == ord('q') or key == 27:
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
```
